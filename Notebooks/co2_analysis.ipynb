###Création d'une fonction pour récupérer les données des fichiers excels

```python
def load_and_concat_excel(file_path):

    # 1. Lire toutes les feuilles
    dfs = pd.read_excel(file_path, sheet_name=None)

    # 2. Récupérer toutes les colonnes existantes
    all_columns = set()
    for df in dfs.values():
        all_columns.update(df.columns.astype(str))

    # 3. Uniformiser les colonnes
    all_columns = sorted(all_columns)
    for name in dfs:
        dfs[name] = dfs[name].reindex(columns=all_columns)

    # 4. Concaténer
    df_final = pd.concat(dfs.values(), ignore_index=True)

    return df_final
```

###1.2.2. Fonction uniformisant les dataframes

```python
def Unif_df(dict_df,dict_col):
  processed_df=[]
  for annee in dict_df.keys():
    df=dict_df[annee].copy()
    df=df.drop(columns=dict_col[annee]['drop'])
    df=df.rename(columns=dict_col[annee]['rename'])
    df['ANNEE']=annee
    processed_df.append(df)
  final_df = pd.concat(processed_df, ignore_index=True)
  final_df.drop_duplicates(inplace=True)
  return final_df
```
